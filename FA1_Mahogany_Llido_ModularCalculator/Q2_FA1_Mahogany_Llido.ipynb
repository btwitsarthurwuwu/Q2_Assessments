```python
def addition(a, b):
    return a + b

def subtraction(a, b):
    return a - b

def multiplication(a, b):
    return a * b

def division(a, b):
    return a / b


first_number = float(input("Enter first number: "))
second_number = float(input("Enter second number: "))

loop = True

while loop:
    operation = input("Enter a number from 1-4: ")

    if operation == "1":
        print(addition(first_number, second_number))
        loop = False
    elif operation == "2":
        print(subtraction(first_number, second_number))
        loop = False
    elif operation == "3":
        print(multiplication(first_number, second_number))
        loop = False
    elif operation == "4":
        if second_number == 0:
            print("Cannot divide by zero!")
        else:
            print(division(first_number, second_number))
        loop = False
    else:
        print("Invalid! Enter a number from 1-4.")
```
